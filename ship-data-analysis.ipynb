```python
df["origin_warehouse"]=df["origin_warehouse"].str.title() #similar like initcap in sq1 first letter and surname first letter will be capital 
```

```python
df.head(30)
```

```python
df["destination_city"]=df["destination_city"].str.title() #
```

```python
df["destination_state"]=df["destination_state"].str.upper() #for upper case
```

```python
df.head(30)
```

```python
df["carrier"]=df["carrier"].str.title()
```

```python
df.head(30)
```

```python
df.isnull().sum() #check in table any null value is present or not 
```

```python
df["destination_city"] = df["destination_city"].fillna(0) #fill null values using fillna()
```

```python
df.isnull().sum()
```

```python

df["ship_date"] = df["ship_date"].fillna(0)
```

```python
df.isnull().sum()
```

```python
df["delivery_date"] = df["delivery_date"].ffill() #it is for forward fill values and other is bfill(backward fil())
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df["damage_reported"] = df["damage_reported"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["freight_cost"] = df["freight_cost"].fillna(0)
```

```python
df.isnull().sum()
```

```python
df.head(30)
```

```python
df["weight_kg"] =  df["weight_kg"].abs()   #for absolute values
```

```python
df["destination_state"] = df["destination_state"].str.strip()
```

```python
df["shipment_status"] = df["shipment_status"].str.title()
```

```python
df.head(30)
```

```python
#PERFORMING EDA
```

```python
df["ship_date"] = pd.to_datetime(df["ship_date"],errors = "coerce") #changing the datatype and any invalid date is present it give Nan
```

```python
df["delivery_date"] = pd.to_datetime(df["delivery_date"],errors = "coerce")
```

```python
df["delivery_days"] = df["delivery_date"] - df["ship_date"] #feature engineering making a new column 
```
